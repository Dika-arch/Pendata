### Apa Itu Missing Values?

Missing values adalah data yang kosong, hilang, atau tidak terisi dalam suatu dataset.

Contoh:

Nilai kosong

NaN (Not a Number)

Null

Contoh:

## Penjelasan Korelasi

Apa Itu Korelasi?

Korelasi adalah ukuran yang menunjukkan seberapa kuat hubungan antara dua variabel numerik.

Nilai korelasi berada di antara:

+1 → hubungan positif sangat kuat

0 → tidak ada hubungan

-1 → hubungan negatif sangat kuat

Jika nilai mendekati:

1 → semakin besar X, semakin besar Y

-1 → semakin besar X, semakin kecil Y

Pada analisis ini digunakan Pearson correlation, yaitu untuk mengukur hubungan linear antar variabel.

📈 Hasil Korelasi Dataset Iris (Pearson)

Nilai utama yang diperoleh:

petal_length ↔ petal_width ≈ 0.96

sepal_length ↔ petal_length ≈ 0.87

sepal_length ↔ petal_width ≈ 0.82

sepal_width ↔ petal_length ≈ -0.42

sepal_width ↔ petal_width ≈ -0.36

sepal_length ↔ sepal_width ≈ -0.11

🔎 Interpretasi
1️⃣ Hubungan Terkuat

Petal_length dan petal_width memiliki korelasi sangat kuat (≈0.96).

Artinya:
Semakin panjang petal, semakin lebar petal.

Variabel ini hampir bergerak bersama.

2️⃣ Hubungan Kuat

Sepal_length dengan petal_length (≈0.87).

Artinya:
Bunga dengan sepal panjang cenderung memiliki petal lebih panjang.

3️⃣ Korelasi Negatif Sedang

Artinya:
Semakin lebar sepal, cenderung petal lebih pendek.

Namun hubungan ini tidak terlalu kuat.

4️⃣ Korelasi Lemah

Sepal_length dengan sepal_width (≈ -0.11).

Artinya:
Hampir tidak ada hubungan antara panjang dan lebar sepal.

**3. Jaccard**

image.png

Jaccard digunakan untuk binary asimetris.

Rumus jarak:

image.png

Dimana:

- q = jumlah 1–1 (keduanya positif)

- r = 1–0

- s = 0–1

- 0–0 tidak dihitung

Intinya:

Fokus pada nilai 1 (kejadian penting)

Dalam kasus :

- Positif = kondisi penting

- Negatif = tidak dihitung jika sama-sama negatif

 **3.1 Matrix Hanya 0 dan 1**

Karena hanya pakai 1 kolom binary.

| Kondisi           | Jarak |
| ----------------- | ----- |
| Positif – Positif | 0     |
| Negatif – Negatif | 0     |
| Positif – Negatif | 1     |

Tidak mungkin muncul 0.5 karena hanya 1 fitur.


 **3.2 Contoh Perhitungan Manual**

Misal:

1.

Pasien 1 = Positif (1)

Pasien 2 = Negatif (0)

Maka:

r+s = 1

q+r+s = 1

d=1/1=1

2.

Pasien 1 = Positif

Pasien 5 = Positif

q = 1

r+s = 0

d=0/1=0

3.

Misal:

Pasien 2 = Negatif

Pasien 4 = Negatif

0–0 diabaikan

Hasil = 0

 **3.3 Interpretasi Matrix**

Lihat pola:

1–5 = 0 → Sama-sama Positif

2–4 = 0 → Sama-sama Negatif

2–6 = 0 → Sama-sama Negatif

Nilai 1 muncul jika satu positif dan satu negatif.

 **3.4 Catatan**

Jaccard distance digunakan untuk atribut binary asimetris (Diabetes_Test). Hasil hanya 0 atau 1 karena hanya satu atribut binary yang digunakan. Nilai 0 menunjukkan kondisi diabetes yang sama, sedangkan nilai 1 menunjukkan perbedaan status. Metode ini lebih tepat untuk data medis karena hanya mempertimbangkan kejadian positif.
